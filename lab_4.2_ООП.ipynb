Вариант 18: Кадровая служба

Базовый класс 1: Employee (оклад)

Дочерние классы 1: Developer (грейд), Designer (портфолио)

Базовый класс 2: Department (назв.)

Дочерний класс 2: ITDepartment (проекты)

Контейнер: Team

Полиморфный метод (бизнес-логика): calculate_bonus() (15% для разраб., 10% для диз.)

```python
class Employee:
    '''  Базовый класс для всех сотрудников. Содержит общие атрибуты и методы.
    ФИО, ID и оклад сотрудников'''
    def __init__(self, employee_fio, employee_id, salary):
        self.employee_fio = employee_fio
        self.employee_id = employee_id
        self.salary = salary

    def __str__(self):
        return f"Сотрудник {self.employee_id}: {self.employee_fio}"

    def calculate_bonus(self):
        """Базовый метод (переопределяется в дочерних классах)"""
        return 0

    def display_info(self):
        """Выводит информацию о сотруднике в консоль"""
        print(f"ID: {self.employee_id}")
        print(f"ФИО: {self.employee_fio}")
        print(f"Оклад: {self.salary:.2f}")
        print(f"Бонус: {self.calculate_bonus():.2f}")

class Developer(Employee):
    ''' Представляет разработчика и его грейд (Junior, Middle, Senior)'''
    def __init__(self, employee_fio, employee_id, salary, grade):
        super().__init__(employee_fio, employee_id, salary)
        self.grade = grade

    def __str__(self):
        parent_str = super().__str__()
        return f"{parent_str} - Грейд: {self.grade}"

    def calculate_bonus(self):
        """Бонус разработчика: 15% от оклада"""
        return self.salary * 0.15

    def display_info(self):
        """Переопределённый метод для разработчика"""
        super().display_info()
        print(f"Грейд: {self.grade}")

class Designer(Employee):
    ''' Представляет дизайнера и количество проектов в его портфолио'''
    def __init__(self, employee_fio, employee_id, salary, portfolio):
        super().__init__(employee_fio, employee_id, salary)
        self.portfolio = portfolio

    def __str__(self):
        parent_str = super().__str__()
        return f"{parent_str} - Портфолио: {self.portfolio} проектов"

    def calculate_bonus(self):
        """Бонус дизайнера: 10% от оклада"""
        return self.salary * 0.10

    def display_info(self):
        """Переопределённый метод для дизайнера"""
        super().display_info()
        print(f"Портфолио: {self.portfolio} проектов")

class Department:
    '''Базовый класс для отделов: содержит название отдела'''
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Отдел: {self.name}"

    def display_info(self):
        """Выводит информацию об отделе"""
        print(f"Название отдела: {self.name}")

    def get_department_type(self):
        """Возвращает тип отдела (для полиморфизма)"""
        return "Обычный"

    def get_bonus_multiplier(self):
        """Множитель бонуса для отдела"""
        return 1.0

class ITDepartment(Department):
    '''Представляет IT-отдел и список проектов'''
    def __init__(self, name, projects):
        super().__init__(name)
        self.projects = projects

    def __str__(self):
        parent_str = super().__str__()
        return f"{parent_str} | IT отдел | Проекты: {', '.join(self.projects)}"

    def display_info(self):
        """Переопределённый метод для IT отдела"""
        super().display_info()
        print(f"Проекты: {', '.join(self.projects)}")

    def get_department_type(self):
        """IT отдел считается VIP-отделом"""
        return "VIP"

    def get_bonus_multiplier(self):
        """Дополнительный множитель бонуса для IT отдела: +10%"""
        return 1.1

class Team:
    """Класс-контейнер для хранения сотрудников и отдела.
    Демонстрирует полиморфизм."""

    def __init__(self, department):
        """ Конструктор контейнера.
        Параметры: department: объект базового класса 2 (Department)"""

        self.department = department  # команда содержит отдел
        self._employees = []          # список сотрудников (Базовый класс 1)

    def add_item(self, employee):
        """ Добавляет сотрудника в команду.
        Параметры: employee: объект базового класса 1 (Employee)"""

        self._employees.append(employee)
        print(f"Добавлен сотрудник: {employee}")

    def remove_item(self, employee_id):
        """Удаляет сотрудника по ID"""
        self._employees = [emp for emp in self._employees if emp.employee_id != employee_id]

    def get_employees(self):
        """Возвращает список сотрудников"""
        return self._employees

    def calculate_total_bonus(self):
        """ Расчёт общей суммы бонусов с учётом типа отдела.
        Полиморфизм:
        1. Для каждого сотрудника вызывается его calculate_bonus()
        2. Применяется множитель от отдела (get_bonus_multiplier())
        3. VIP-отдел (ITDepartment) даёт дополнительный бонус
        """
        total = 0
        for emp in self._employees:
            # Полиморфный вызов: у каждого объекта своя реализация calculate_bonus()
            bonus = emp.calculate_bonus()
            # Применяется множитель отдела
            total += bonus * self.department.get_bonus_multiplier()
        return total

    def display_report(self):
        """Выводит детализированный отчёт о команде"""
        print()
        print(f"Детализированный отчёт о команде {self.department}:")
        print()

        print(f"{self.department}")
        print(f"Тип отдела: {self.department.get_department_type()}")
        print(f"Множитель бонуса: {self.department.get_bonus_multiplier()}x")

        print(f"\n Сотрудники ({len(self._employees)} чел.):")
        print()

        for emp in self._employees:
            emp.display_info()
            print()

        print(f"Итоговая сумма бонусов: {self.calculate_total_bonus():.2f}")
        print()



# ДЕМОНСТРАЦИЯ РАБОТЫ:
def main():
    print()
    print("Система управления персоналом:")
    print()

    # Сотрудники разработчики
    dev1 = Developer("Иванов И.И.", 101, 100000, "Senior")
    dev2 = Developer("Петров П.П.", 102, 80000, "Middle")
    dev3 = Developer("Муравьёв П.И.", 103, 60000, "Junior")
    dev4 = Developer("Сидоров С.С.", 104, 120000, "Team Lead")

    # Сотрудники-дизайнеры
    des1 = Designer("Кузнецова А.А.", 201, 90000, 15)
    des2 = Designer("Васильева Е.Е.", 202, 85000, 8)


    # Обычный отдел
    regular_dept = Department("Бухгалтерия")
    print(f"Создан обычный отдел: {regular_dept}")

    # VIP-отдел (ITDepartment)
    it_projects = ["Проект Альфа", "Проект Бета", "Проект Гамма"]
    it_dept = ITDepartment("Технологический", it_projects)
    print(f"Создан VIP отдел: {it_dept}")

    print("\nСоздание команд (контейнеров):")
    print()

    # Команда 1: Обычный отдел
    team_regular = Team(regular_dept)
    team_regular.add_item(dev1)
    team_regular.add_item(dev4)
    team_regular.add_item(des1)

    # Команда 2: VIP-отдел (IT)
    team_vip = Team(it_dept)
    team_vip.add_item(dev2)
    team_vip.add_item(dev3)
    team_vip.add_item(des2)

    print("\nДемонстрация полиморфизма:")

    print("\nПолиморфизм метода calculate_bonus():")
    print("(разная логика для разных типов сотрудников)")

    print(f"\nРазработчик - {dev1.employee_fio}: бонус {dev1.calculate_bonus():.2f} (15%)")
    print(f"Дизайнер - {des1.employee_fio}: бонус {des1.calculate_bonus():.2f} (10%)")

    print("\nПолиморфизм метода get_bonus_multiplier():")
    print(f"Отдел бухгалтерии (обычный): множитель {regular_dept.get_bonus_multiplier()}x")
    print(f"IT отдел (VIP): множитель {it_dept.get_bonus_multiplier()}x")


    print("\nИтоговые отчёты")

    # Отчёт для команды обычного отдела
    team_regular.display_report()

    # Отчёт для VIP-команды
    team_vip.display_report()

    print()
    print("Сравнительный анализ:")

    total_regular = team_regular.calculate_total_bonus()
    total_vip = team_vip.calculate_total_bonus()

    print(f"\nОбщий бонус обычной команды: {total_regular:.2f}")
    print(f"Общий бонус VIP-команды: {total_vip:.2f}")
    print(f"Разница: {total_vip - total_regular:.2f}")

    print("\nРабота полиморфизма подтверждена:")
    print("   - Тип сотрудника имеет свою формулу бонуса")
    print("   - Тип отдела - свой множитель")


if __name__ == "__main__":
  try:
    main()
  except TypeError as e:
    print(f"Ошибка: неверный тип данных - {e}")
  except ValueError as e:
    print(f"Ошибка: неверное значение - {e}")
```
