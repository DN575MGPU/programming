```python
# Задание task_02_18.
#
# Выполнил: Новожилкина Д.В.
# Группа: ЦИБ-251



sentence = input('Введите предложение: ')

count_gl = 0  # Кол-во гласных
count_sogl = 0  # Кол-во согласных

vowels = set('аоуеиюяэыАОУЕИЯЮЫЭ')
consonants = set('бвгджзйклмнпрстфхцчшщБВГДЖЗЙКЛМНПРСТФХЦЧШЩ')

for char in sentence:
  if char == ' ':
    continue
  elif char in vowels:
    count_gl += 1
  else:
    count_sogl += 1

print(f'Количество букв в предложении: гласных - {count_gl}, согласных - {count_sogl}')
```
