найдите потенциальные источники ошибок (укажите номера строк в строке документации);

используя конструкцию try добавьте в код обработку соответствующих исключений.

```python
# Задание task_03_03_07.
# Выполнил: Новожилкина Д.В.
# Группа: ЦИБ-251



"""
Потенциальные ошибки в строках - 28, 29: Пользователь может ввести не целое число,
отрицательную степень или слишком большую степень, которая не поддерживается программой.
"""


def power(x, y=2):

    """Вернуть x^y. Поддерживает только целые неотрицательные степени."""

    if y < 0:
        raise ValueError("Отрицательная степень не поддерживается")
    if y >= 1000:
        raise ValueError('Слишком большая положительная степень не поддерживается')
    if y == 0:
        return 1
    else:
        return x * power(x, y - 1)


try:
    x_str = input("x = ")
    y_str = input("y = ")

    # Проверяет валидность ввода.
    try:
        x = int(x_str)
    except ValueError:
        raise TypeError("Основание x должно быть целым числом")

    try:
        y = int(y_str)
    except ValueError:
        raise TypeError("Степень y должна быть целым числом")

    print(power(x, y))

except ValueError as e:
    print(f"Ошибка ввода: {e}")
except TypeError as e:
    print(f"Ошибка типа: {e}")
```
