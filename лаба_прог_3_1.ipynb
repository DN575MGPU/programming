```python
# Задание task_03_01_01.
#
# Выполнила: Новожилкина Д. В.
# Группа: ЦИБ-251

import math
def sgn(x):
    """Вернуть результат сигнум-функции.
    Параметры:
        - x (float): аргумент x.
    Результат:
        float: значение функции.
    """
    if x > 0:
      return 1
    elif x == 0:
      return 0
    return -1

x = float(input("Введите x: "))
y = float(input("Введите y: "))

numerator = sgn(x) + y ** 2               # числитель
denominator = sgn(y) - math.sqrt(abs(x))  # знаменатель
z = numerator / denominator
try:
  denominator != 0
except ValueError:
  print("Ошибка: знаменатель равен нулю")

print("Ответ:", round(z, 2))

# --------------
# Пример вывода:
#
# Введите x: -9
# Введите y: 0
# Ответ: 0.33
```
